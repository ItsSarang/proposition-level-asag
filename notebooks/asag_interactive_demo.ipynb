### Educational & Methodological Overview

#### Classical Base ASAG Pipeline (System 2)
The Base ASAG system evaluates student answers through a transparent, interpretable pipeline of classical NLP stages:
$$\text{Raw Text} \xrightarrow{\text{Stage 3}} \text{Preprocessing} \xrightarrow{\text{Stage 4}} \text{Parsing} \xrightarrow{\text{Stage 5}} \text{Triples} \xrightarrow{\text{Stage 6}} \text{Semantics} \xrightarrow{\text{Stage 7}} \text{Matching} \xrightarrow{\text{Stage 8}} \text{Entailment} \xrightarrow{\text{Stage 9}} \text{Scoring}$$

#### Why Base ASAG Outperforms Surface Bag-of-Words (TF-IDF):
1. **Role Preservation:** Subjects match subjects, predicates match predicates, objects match objects. Permuted words do not falsely inflate scores.
2. **Polarity Sensitivity:** Explicit negation (*"never uniquely identifies"*) triggers `CONTRADICTS` ($0.00$ credit).
3. **Traceability:** Every decision is backed by inspectable tokens, lemmas, synsets, and alignment scores.

#### The Core Research Limitation & Motivation for Proposed Extension:
In classical entailment ASAG (Basak et al., 2019), scoring evaluates **reference answer coverage**:
$$\text{normalized\_score} = \frac{\sum \text{credit}(\text{Reference Proposition}_i)}{N_{\text{ref}}}$$

Because extra information provided by the student is outside the reference proposition set, **it is assigned zero score effect ($0.00$)**.  
Consequently, Base ASAG cannot distinguish:
- **Correct Additional Information** (e.g. *"cannot contain NULL"*) $\rightarrow$ Score: $1.00$
- **Irrelevant Additional Information** (e.g. *"SQL was developed in 1970s"*) $\rightarrow$ Score: $1.00$
- **Incorrect Additional Information** (e.g. *"can contain duplicate values"*) $\rightarrow$ Score: $1.00$

This exact limitation motivates the central research question of the project:
> *"Can an interpretable proposition-level extension of a classical semantic/dependency-based ASAG system evaluate unmatched student information as correct additional, irrelevant, or incorrect information, and does incorporating this evaluation improve agreement with human grading?"*
